```python
# formatting the price_in_million column for better code undestanding output

def formatted_price(value_millions):
    """
    converts prices from million to either billion or trillion by using 1000 count for billion spotting and 1_000_000 count for trillion
    as 1000 million makes a billion and 1_000_000 million makes a trillion
    after which if it found any output as integer convert it into float for better consistency
    """
    if value_millions >= 1_000_000:           # 1 trillion Naira or more
        trillions = value_millions / 1_000_000
        return f"{trillions:.1f}T" if trillions.is_integer() else f"{trillions:.1f}T"
    elif value_millions >= 1_000:             # 1 billion or more
        billions = value_millions / 1_000
        return f"{billions:.1f}B" if billions.is_integer() else f"{billions:.1f}B"
    else:
        return f"{value_millions:.1f}M" if value_millions.is_integer() else f"{value_millions:.1f}M"
```

```python
# Apply the formatting

df['price_formatted'] = df['price_million_ngn'].apply(formatted_price)
```

```python
# Inserting it right after price_in_million for perfect column order

col_order = df.columns.get_loc('price_million_ngn') + 1
df.insert(col_order, 'price_formatted', df.pop('price_formatted'))
```

```python
# Now time to convert to real datetime

def date_conversion_parser(text):
    if pd.isna(text):
        return pd.NaT                         # not a time
    text = str(text).lower()
    num = int(re.search(r'(\d+)', text).group(1)) if re.search(r'\d+', text) else 0
    
    if 'minute' in text or 'now' in text:  return SCRAPE_DATE
    if 'hour'   in text:  return SCRAPE_DATE - timedelta(hours=num)
    if 'day'    in text:  return SCRAPE_DATE - timedelta(days=num)
    if 'week'   in text:  return SCRAPE_DATE - timedelta(weeks=num)
    if 'month'  in text:  return SCRAPE_DATE - timedelta(days=num*30)
    if 'year'   in text:  return SCRAPE_DATE - timedelta(days=num*365)
    return SCRAPE_DATE  
```

```python
df['listing_date'] = df['listing_date'].apply(date_conversion_parser)
```

```python
def price_format(price_ngn):
    if pd.isna(price_ngn):
        return ""
    
    price_ngn = float(price_ngn)
    
    if price_ngn >= 1_000_000:
        value = price_ngn / 1_000_000
        if value >= 10:
            return f"₦{value:.1f}M".replace('.0M', 'M')
        else:
            formatted = f"₦{value:.2f}M"
            return formatted.rstrip('0').rstrip('.') if '.' in formatted else formatted
    
    elif price_ngn >= 100_000:
        value = price_ngn / 1_000
        return f"₦{value:.0f}K"
    
    else:  
        return f"₦{price_ngn:,.0f}"
```

```python
# Now time to convert to real datetime

def date_conversion_parser(text):
    if pd.isna(text):
        return pd.NaT                         # not a time
    text = str(text).lower()
    num = int(re.search(r'(\d+)', text).group(1)) if re.search(r'\d+', text) else 0
    
    if 'minute' in text or 'now' in text:  return SCRAPE_DATE
    if 'hour'   in text:  return SCRAPE_DATE - timedelta(hours=num)
    if 'day'    in text:  return SCRAPE_DATE - timedelta(days=num)
    if 'week'   in text:  return SCRAPE_DATE - timedelta(weeks=num)
    if 'month'  in text:  return SCRAPE_DATE - timedelta(days=num*30)
    if 'year'   in text:  return SCRAPE_DATE - timedelta(days=num*365)
    return SCRAPE_DATE  
```

```python
df_rent['listing_date'] = df_rent['listing_date'].apply(date_conversion_parser)
```
